<h2 style="color:DodgerBlue">Описание:</h2>

----

Проект для управления растениями в саду. Базовый класс `Plant` (Растение), от него наследуются `Flower` (Цветок) и `Tree` (Дерево). Метод `Bloom()` (Цветение) объявлен как `virtual` в базовом классе и переопределён через `override` в наследниках — это демонстрация полиморфизма.

| Класс | Поля | Методы |
|---|---|---|
| `Plant` (Растение) | `Name`, `Type`, `Height` | `Bloom()`, `Grow(float growth)` |
| `Flower` (Цветок) | `Color`, `PetalCount` | `Bloom()` (override), `Wither()` |
| `Tree` (Дерево) | `Age`, `LeafType` | `Bloom()` (override), `GrowOlder(int years)` |


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"{Name} готовится зацвести");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"{Name} подросло на {growth} см, новая высота: {Height} см");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount) : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wither()
    {
        Console.WriteLine($"{Name} постепенно увядает");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType) : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"{Name} стало старше на {years} лет, текущий возраст: {Age}");
    }
}

Flower tulip = new Flower("Тюльпан", 18, "Жёлтый", 6);
Tree maple = new Tree("Клён", 420, 12, "Лиственные");

Console.WriteLine($"{tulip.Name} ({tulip.Type}): цвет {tulip.Color}, лепестков {tulip.PetalCount}, высота {tulip.Height} см");
Console.WriteLine($"{maple.Name} ({maple.Type}): листья {maple.LeafType}, возраст {maple.Age} лет, высота {maple.Height} см");
Console.WriteLine();

tulip.Bloom();
tulip.Grow(3.5f);
tulip.Wither();
Console.WriteLine();

maple.Bloom();
maple.Grow(25f);
maple.GrowOlder(1);
Console.WriteLine();

List<Plant> garden = new List<Plant>();
garden.Add(tulip);
garden.Add(maple);

foreach (Plant plant in garden)
{
    plant.Bloom();
}
```
