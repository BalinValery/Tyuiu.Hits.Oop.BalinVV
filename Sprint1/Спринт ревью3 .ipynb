```
public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public string Manufacturer { get; set; }
    public decimal Price { get; set; }

    protected Component(string name, string manufacturer, decimal price)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, производитель: {Manufacturer}, цена: {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Name} ({Manufacturer}), {Price} руб.";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double FrequencyGHz { get; set; }

    public CPU(string name, string manufacturer, decimal price, int cores, double frequencyGHz) : base(name, manufacturer, price)
    {
        Cores = cores;
        FrequencyGHz = frequencyGHz;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Тип: процессор, ядер: {Cores}, частота: {FrequencyGHz} ГГц");
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен в сокет материнской платы");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Cores == b.Cores && a.FrequencyGHz == b.FrequencyGHz;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Cores, FrequencyGHz);
    }
}

public class RAM : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public string MemoryType { get; set; }

    public RAM(string name, string manufacturer, decimal price, int capacityGB, string memoryType) : base(name, manufacturer, price)
    {
        CapacityGB = capacityGB;
        MemoryType = memoryType;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Тип: оперативная память, объём: {CapacityGB} ГБ, стандарт: {MemoryType}");
    }

    public void Install()
    {
        Console.WriteLine($"Модуль памяти {Name} установлен в слот DIMM");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.CapacityGB == b.CapacityGB && a.MemoryType == b.MemoryType;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, CapacityGB, MemoryType);
    }
}

public class GPU : Component, IInstallable
{
    public int VideoMemoryGB { get; set; }

    public GPU(string name, string manufacturer, decimal price, int videoMemoryGB) : base(name, manufacturer, price)
    {
        VideoMemoryGB = videoMemoryGB;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Тип: видеокарта, видеопамять: {VideoMemoryGB} ГБ");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлена в слот PCI-Express");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.VideoMemoryGB == b.VideoMemoryGB;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, VideoMemoryGB);
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public int Rpm { get; set; }

    public HDD(string name, string manufacturer, decimal price, int capacityGB, int rpm) : base(name, manufacturer, price)
    {
        CapacityGB = capacityGB;
        Rpm = rpm;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Тип: жёсткий диск, объём: {CapacityGB} ГБ, скорость вращения: {Rpm} об/мин");
    }

    public void Install()
    {
        Console.WriteLine($"Жёсткий диск {Name} подключён к разъёму SATA");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.CapacityGB == b.CapacityGB && a.Rpm == b.Rpm;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, CapacityGB, Rpm);
    }
}

public class Computer
{
    public string Name { get; set; }
    private List<Component> components = new List<Component>();

    public List<Component> Components
    {
        get { return components; }
    }

    public Computer(string name)
    {
        Name = name;
    }

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        component.Install();
        components.Add(component);
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удалён из компьютера {Name}");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден в компьютере {Name}");
        }
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (Component component in components)
        {
            total += component.Price;
        }
        return total;
    }

    // Перегрузка метода: стоимость с учётом скидки в процентах
    public decimal GetTotalPrice(decimal discountPercent)
    {
        return GetTotalPrice() * (100 - discountPercent) / 100;
    }

    public void PrintConfiguration()
    {
        Console.WriteLine($"Конфигурация компьютера \"{Name}\":");
        foreach (Component component in components)
        {
            Console.WriteLine("- " + component);
        }
    }
}

public static class ComputerFactory
{
    public static Computer CreateCustom<TCpu, TRam, TGpu, THdd>(string name, TCpu cpu, TRam ram, TGpu gpu, THdd hdd)
        where TCpu : CPU
        where TRam : RAM
        where TGpu : GPU
        where THdd : HDD
    {
        Computer computer = new Computer(name);
        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(gpu);
        computer.AddComponent(hdd);
        return computer;
    }

    public static Computer CreateOfficeComputer()
    {
        return CreateCustom(
            "Офисный ПК",
            new CPU("Core i3-12100", "Intel", 9500m, 4, 3.3),
            new RAM("Kingston 8GB", "Kingston", 2500m, 8, "DDR4"),
            new GPU("Встроенная графика UHD 730", "Intel", 0m, 1),
            new HDD("Barracuda 1TB", "Seagate", 3800m, 1000, 7200));
    }

    public static Computer CreateGamingComputer()
    {
        return CreateCustom(
            "Игровой ПК",
            new CPU("Ryzen 7 7800X3D", "AMD", 36000m, 8, 4.2),
            new RAM("Fury Beast 32GB", "Kingston", 9500m, 32, "DDR5"),
            new GPU("GeForce RTX 4070", "NVIDIA", 58000m, 12),
            new HDD("WD Blue 2TB", "Western Digital", 5200m, 2000, 7200));
    }
}

CPU cpu1 = new CPU("Core i5-12400", "Intel", 15000m, 6, 2.5);
CPU cpu2 = new CPU("Core i5-12400", "Intel", 15500m, 6, 2.5);
CPU cpu3 = new CPU("Ryzen 5 5600", "AMD", 12000m, 6, 3.5);
RAM ram = new RAM("Crucial 16GB", "Crucial", 4500m, 16, "DDR4");
GPU gpu = new GPU("GeForce GTX 1660", "NVIDIA", 22000m, 6);
HDD hdd = new HDD("Toshiba P300 1TB", "Toshiba", 3900m, 1000, 7200);


Component[] parts = { cpu1, ram, gpu, hdd };
foreach (Component part in parts)
{
    part.GetInfo();
    Console.WriteLine("ToString: " + part);
}
Console.WriteLine();


Console.WriteLine($"cpu1 == cpu2 (те же характеристики): {cpu1 == cpu2}");
Console.WriteLine($"cpu1 != cpu3 (разные процессоры): {cpu1 != cpu3}");
Console.WriteLine($"cpu1.Equals(ram): {cpu1.Equals(ram)}");
Console.WriteLine();


Computer myPc = new Computer("Сервисный ПК");
myPc.AddComponent(cpu1);
myPc.AddComponent(ram);
myPc.AddComponent(gpu);
myPc.AddComponent(hdd);
myPc.PrintConfiguration();
Console.WriteLine($"Общая стоимость: {myPc.GetTotalPrice()} руб., со скидкой 10%: {myPc.GetTotalPrice(10m)} руб.");
Console.WriteLine();


myPc.RemoveComponent(gpu);
myPc.RemoveComponent(gpu);
Console.WriteLine($"Стоимость после удаления: {myPc.GetTotalPrice()} руб.");
Console.WriteLine();


Computer office = ComputerFactory.CreateOfficeComputer();
Console.WriteLine($"Стоимость \"{office.Name}\": {office.GetTotalPrice()} руб.");
Computer gaming = ComputerFactory.CreateGamingComputer();
Console.WriteLine($"Стоимость \"{gaming.Name}\": {gaming.GetTotalPrice()} руб.");
```
