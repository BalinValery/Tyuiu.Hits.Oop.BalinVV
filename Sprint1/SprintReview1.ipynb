<h1 style="color:DodgerBlue">Спринт ревью #1</h1>

<h2 style="color:DodgerBlue">Описание:</h2>

----

Классы `Product` (Товар), `Seller` (Продавец), `Store` (Магазин). Поля закрыты модификаторами доступа `private`, `protected`, `internal`, `public`, доступ к ним через свойства с геттерами и сеттерами. От каждого класса создан один экземпляр, объекты взаимодействуют между собой: магазин хранит продавцов, продавец хранит и продает товары.

Карта спринта 1

Product (товар)
  private    name          название
  private    price         цена
  private    quantity      количество
  protected  category      категория
  internal   description   описание

  GetTotalPrice()          посчитать общую стоимость (цена × количество)
  UpdateQuantity(amount)   изменить количество на amount (+ или −)
  GetProductInfo()         рассказать о товаре

Seller (продавец)
  private    name          имя
  private    employeeId    табельный номер
  protected  salary        зарплата
  internal   contactInfo   контакты
  private    products      список его товаров

  AddProduct(product)              добавить товар в свой список
  SellProduct(product, quantity)   продать: уменьшить количество на складе
  GetSellerInfo()                  рассказать о себе
  
Store (магазин)
  private    storeName     название
  private    location      адрес
  public     storeHours    часы работы
  private    sellers       список продавцов

  AddSeller(seller)    добавить продавца
  ListProducts()       показать все товары всех продавцов
  GetStoreInfo()       рассказать о себе


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set
        {
            if (value >= 0)
                price = value;
            else
                Console.WriteLine("Цена не может быть отрицательной!");
        }
    }

    public int Quantity
    {
        get { return quantity; }
        set
        {
            if (value >= 0)
                quantity = value;
            else
                Console.WriteLine("Количество не может быть отрицательным!");
        }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        this.description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity = quantity + amount;
    }

    public string GetProductInfo()
    {
        return $"Товар: {name}, категория: {category}, цена: {price} руб., количество: {quantity} шт., описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public List<Product> Products
    {
        get { return products; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        this.contactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Продавец {name} добавил товар {product.Name}");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"Продавец {name} продал {quantity} шт. товара {product.Name}, осталось {product.Quantity} шт.");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара {product.Name} на складе");
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {name}, ID: {employeeId}, зарплата: {salary} руб., контакты: {contactInfo}, товаров: {products.Count}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers = new List<Seller>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"В магазин {storeName} добавлен продавец {seller.Name}");
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в магазине {storeName}:");
        foreach (Seller seller in sellers)
        {
            foreach (Product product in seller.Products)
            {
                Console.WriteLine("- " + product.GetProductInfo());
            }
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, адрес: {location}, часы работы: {storeHours}, продавцов: {sellers.Count}";
    }
}

Product apple = new Product("Яблоки", 100m, 10, "Фрукты", "Яблоки зеленые");
Seller seller = new Seller("Иван Иванов", "EMP001", 40000m, "ivan@mail.ru");
Store store = new Store("Теплица", "г. Тюмень, ул. 50 лет октября, 1", "9:00 - 21:00");

Console.WriteLine(apple.GetProductInfo());
Console.WriteLine($"Общая стоимость: {apple.GetTotalPrice()} руб.");
Console.WriteLine();

store.AddSeller(seller);
seller.AddProduct(apple);
Console.WriteLine();

seller.SellProduct(apple, 3);
seller.SellProduct(apple, 20);
apple.UpdateQuantity(5);
Console.WriteLine($"После поставки: {apple.Quantity} шт., общая стоимость: {apple.GetTotalPrice()} руб.");
Console.WriteLine();

apple.Price = 100m;
apple.Price = -100m;
Console.WriteLine();

Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine(store.GetStoreInfo());
store.ListProducts();
```
