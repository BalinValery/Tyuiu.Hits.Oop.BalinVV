<h2 style="color:DodgerBlue">Описание:</h2>

----

Структура компьютерного сервиса. Базовый класс `Employee` (Сотрудник), от него наследуется `Manager` (Менеджер), а от него — `ProjectManager` (Проектный менеджер): это простое и сложное наследование. Множественное наследование в C# невозможно, поэтому `ProjectManager` дополнительно реализует два интерфейса: `IReportable` (формирование отчёта) и `ITeamLeader` (состав команды). Метод `PrintInfo()` объявлен как `virtual` в `Employee` и переопределён (`override`) в `Manager` и `ProjectManager`.

| Тип | Поля / свойства | Методы |
|---|---|---|
| `IReportable` (интерфейс) | — | `GenerateReport()` |
| `ITeamLeader` (интерфейс) | — | `GetTeamMembers()` |
| `Employee` (Сотрудник) | `Name`, `Position`, `Salary` | `PrintInfo()` |
| `Manager` : `Employee` | `Department` | `PrintInfo()` (override) |
| `ProjectManager` : `Manager`, `IReportable`, `ITeamLeader` | `ProjectName`, `teamMembers` | `PrintInfo()` (override), `AddTeamMember()`, `GenerateReport()`, `GetTeamMembers()` |


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, должность: {Position}, зарплата: {Salary} руб.");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department) : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Руководит отделом: {Department}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    private List<string> teamMembers = new List<string>();

    public ProjectManager(string name, string position, double salary, string department, string projectName) : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    public void AddTeamMember(string member)
    {
        teamMembers.Add(member);
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отвечает за проект: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"Отчёт по проекту \"{ProjectName}\": руководитель {Name} ({Department}), участников в команде: {teamMembers.Count}";
    }

    public string[] GetTeamMembers()
    {
        return teamMembers.ToArray();
    }
}

Employee master = new Employee("Алексей Смирнов", "Мастер по ремонту ПК", 55000);
Manager manager = new Manager("Ирина Козлова", "Руководитель сервисного центра", 90000, "Отдел ремонта ноутбуков");
ProjectManager projectManager = new ProjectManager("Дмитрий Орлов", "Проектный менеджер", 110000, "Отдел корпоративных клиентов", "Модернизация компьютерного парка для офиса клиента");

projectManager.AddTeamMember("Алексей Смирнов (мастер по ремонту ПК)");
projectManager.AddTeamMember("Елена Волкова (сетевой инженер)");
projectManager.AddTeamMember("Павел Лебедев (специалист по восстановлению данных)");

Console.WriteLine("PrintInfo() для каждого объекта");
master.PrintInfo();
Console.WriteLine();
manager.PrintInfo();
Console.WriteLine();
projectManager.PrintInfo();
Console.WriteLine();

Console.WriteLine("Методы интерфейсов в ProjectManager");
IReportable reportable = projectManager;
Console.WriteLine(reportable.GenerateReport());
Console.WriteLine();

ITeamLeader teamLeader = projectManager;
Console.WriteLine("Состав команды:");
foreach (string member in teamLeader.GetTeamMembers())
{
    Console.WriteLine("- " + member);
}
Console.WriteLine();

Console.WriteLine("Все сотрудники через общий тип Employee ");
List<Employee> staff = new List<Employee>();
staff.Add(master);
staff.Add(manager);
staff.Add(projectManager);

foreach (Employee employee in staff)
{
    employee.PrintInfo();
    Console.WriteLine();
}
```
